```python
@dataclass
class City:
    name: str
    population: int = 0
    tags: list[str] = field(default_factory=list)

    @property
    def label(self) -> str:
        return f"{self.name} ({self.population:,})"

    @classmethod
    def tokyo(cls) -> "City":
        return cls("Tokyo", 14_000_000)

    def rename(self, new_name: str) -> None:
        self.name = new_name

    def __repr__(self) -> str:
        return self.label


class Megacity(City):
    def __init__(self, name: str) -> None:
        super().__init__(name, 10_000_000)
        self.visits = 0


osaka = Megacity("Osaka")
osaka.rename("Ōsaka")
print(City.tokyo(), osaka, osaka.visits)
```

| <span style="color:#7c3aed;font-family:monospace">async</span> | <span style="color:#7c3aed;font-family:monospace">#7c3aed</span> | <code>no semantic token · storage.type.function.async</code> |

| <span style="color:#0e7490;font-family:monospace">ValueError  KeyError</span> | <span style="color:#0e7490;font-family:monospace">#0e7490</span> | <code>class.builtin · support.type.exception</code> |

```python
counter = 0


def bump() -> None:
    global counter
    counter += 1


def describe(value: object) -> str:
    match value:
        case int(n) if n > 0:
            return "positive int"
        case [first_item, *rest]:
            return f"list starting with {first_item}"
        case _:
            return "something else"


try:
    if (n := len(os.sep)) > 0:
        raise ValueError(f"separator length {n}")
except (ValueError, KeyError) as err:
    print("caught:", err)
finally:
    for _ in range(2):
        bump()


async def fetch() -> str:
    return "done"

print(describe(5), describe([1, 2]), await fetch(), counter)
```
